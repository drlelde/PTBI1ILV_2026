• The number must be divisible by five.
• If the number is greater than or equal to 150, then skip it.
• If the number is greater than 500, then stop the loop.
Example input:
• Python: [12, 75, 150, 182, 145, 525, 50]

Problem 8: Write code that reverses a given string number s. For example “stressed“ should result in “desserts“.
Similarly, write code that reverses a given integer n. For example n = 76542 should result in m = 24567.

```python
n = 76542
m = 0                    # the reversed number we're building
while n > 0:
    last_digit = n % 10      # grab the LAST digit
    m = m * 10 + last_digit  # tack it onto m
    n = n // 10              # chop the last digit off
print(m)
```
